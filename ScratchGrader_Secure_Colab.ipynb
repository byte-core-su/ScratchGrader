```python
import os
from google.colab import userdata

required = ['NGROK_AUTHTOKEN', 'NGROK_STATIC_DOMAIN', 'ADMIN_TOKEN']
optional = ['NGROK_API_KEY', 'NGROK_REMOTE_RECOVERY', 'FIREBASE_ENABLED', 'FIREBASE_PROJECT_ID', 'FIREBASE_SERVICE_ACCOUNT_JSON', 'CORS_ALLOWED_ORIGINS', 'MAX_CONCURRENT_GRADES', 'MAX_QUEUED_GRADES']
for name in required + optional:
    try:
        value = userdata.get(name)
        if value:
            os.environ[name] = value
    except Exception:
        pass

missing = [name for name in required if not os.environ.get(name)]
if missing:
    raise RuntimeError('請先在 Colab Secrets 設定：' + ', '.join(missing))

if os.environ.get('FIREBASE_ENABLED', '').lower() in ('1', 'true', 'yes', 'on'):
    firebase_missing = [name for name in ['FIREBASE_PROJECT_ID', 'FIREBASE_SERVICE_ACCOUNT_JSON'] if not os.environ.get(name)]
    if firebase_missing:
        raise RuntimeError('已啟用 Firebase，但缺少：' + ', '.join(firebase_missing))
```
